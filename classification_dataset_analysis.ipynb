```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

# 1. Load the dataset
data = load_breast_cancer(as_frame=True)
df = data.frame
df.to_csv("breast_cancer.csv", index=False)

# 2. Dataset overview
print("Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print(df.dtypes.value_counts())

# 3. Features (X) and target (y)
X = df.drop(columns=["target"])
y = df["target"]
print("Number of features:", X.shape[1])
print("Feature names:", list(data.feature_names))
print("X shape:", X.shape, "| y shape:", y.shape)

# 4. Target classes
print("Class names:", list(data.target_names))      # ['malignant', 'benign']
print("Unique target values:", sorted(y.unique()))  # [0, 1]
print(y.value_counts())
print(y.value_counts(normalize=True).round(3))

# 5. Binary or multiclass?
n_classes = y.nunique()
print("Number of classes:", n_classes)
print("Problem type:", "Binary classification" if n_classes == 2 else "Multiclass classification")

# Safety check: target must not be in X
assert "target" not in X.columns
```
